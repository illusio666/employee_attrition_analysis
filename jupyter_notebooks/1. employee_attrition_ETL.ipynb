### While there are no empty cells that need to be filled, there are still some improvements to the data that can be made:

1) three columns have constant values (employee no, hours, Over 18) and so those can be removed for cleanliness as they add no value
2) employee number can be removed for the purpose of analysis, albeit this would need to be re-aligned at a later date if HR wished to deploy bespoke retention strategies to different clusters of employees (TBC once analysis done)
3) while there are several fields with zeros for values, these are legitimate values for those fields e.g. number of years since last promotion. Therefore these should be left unchanged.
4) some values are numerical values stored as categorical, when in fact they represent ordinal sequences e.g. education level, satisfaction levels, etc. So those will be amended display an ordinal, descriptive value. While there is no description available for the data sets of exactly what the values represent, an assumption of descriptions will be made; while the descriptions may not be exact, any correlation between a value and another data field will not be affected.
5) There may be some value for subsequent analysis to address relative tenure and income (possible income and job stagnation). Therefore we will add columns for:

- ratio of years in current role to years at company
- ratio of years with current manager to years at company
- ratio of monthly income to years at company
- ratio of monthly income to total working years

```python
## Drop the 4 columns that are not useful for analysis
df.drop(columns=['EmployeeCount', 'EmployeeNumber', 'Over18','StandardHours'],inplace=True)
```

Now we need to convert the 'level' values into sequenced, descriptive values.

```python
#firstly, ensure the fields are read as integers
df['Education'] = df['Education'].astype(int)
df['EnvironmentSatisfaction'] = df['EnvironmentSatisfaction'].astype(int)
df['JobInvolvement'] = df['JobInvolvement'].astype(int)
df['JobSatisfaction'] = df['JobSatisfaction'].astype(int)
df['PerformanceRating'] = df['PerformanceRating'].astype(int)
df['RelationshipSatisfaction'] = df['RelationshipSatisfaction'].astype(int)
df['WorkLifeBalance'] = df['WorkLifeBalance'].astype(int)   
df.dtypes
```

```python
# Define mapping dictionaries with integer keys
education_map = {
    1: '1-Below College',
    2: '2-College',
    3: '3-Bachelor',
    4: '4-Master',
    5: '5-Doctor'
}
env_satisfaction_map = {
    1: '1-Low',
    2: '2-Medium',
    3: '3-High',
    4: '4-Very High'
}
job_involvement_map = {
    1: '1-Low',
    2: '2-Medium',
    3: '3-High',
    4: '4-Very High'
}
job_satisfaction_map = {
    1: '1-Low',
    2: '2-Medium',
    3: '3-High',
    4: '4-Very High'
}
performance_rating_map = {
    1: '1-Low',
    2: '2-Good',
    3: '3-Excellent',
    4: '4-Outstanding'
}
relationship_satisfaction_map = {
    1: '1-Low',
    2: '2-Medium',
    3: '3-High',
    4: '4-Very High'
}
work_life_balance_map = {
    1: '1-Bad',
    2: '2-Good',
    3: '3-Better',
    4: '4-Best'
}

# Apply mappings
df['Education'] = df['Education'].map(education_map)
df['EnvironmentSatisfaction'] = df['EnvironmentSatisfaction'].map(env_satisfaction_map)
df['JobInvolvement'] = df['JobInvolvement'].map(job_involvement_map)
df['JobSatisfaction'] = df['JobSatisfaction'].map(job_satisfaction_map)
df['PerformanceRating'] = df['PerformanceRating'].map(performance_rating_map)
df['RelationshipSatisfaction'] = df['RelationshipSatisfaction'].map(relationship_satisfaction_map)
df['WorkLifeBalance'] = df['WorkLifeBalance'].map(work_life_balance_map)

# Define category orders
ordinal_orders = {
    'Education': ['1-Below College', '2-College', '3-Bachelor', '4-Master', '5-Doctor'],
    'EnvironmentSatisfaction': ['1-Low', '2-Medium', '3-High', '4-Very High'],
    'JobInvolvement': ['1-Low', '2-Medium', '3-High', '4-Very High'],
    'JobSatisfaction': ['1-Low', '2-Medium', '3-High', '4-Very High'],
    'PerformanceRating': ['1-Low', '2-Good', '3-Excellent', '4-Outstanding'],
    'RelationshipSatisfaction': ['1-Low', '2-Medium', '3-High', '4-Very High'],
    'WorkLifeBalance': ['1-Bad', '2-Good', '3-Better', '4-Best']
}

# Convert to ordered categorical
for col, order in ordinal_orders.items():
    df[col] = pd.Categorical(df[col], categories=order, ordered=True)

# Preview
df.head(5)
```

Now to add the four 'ratio' values identified above, to assist the analysis to identify possible income or job stagnation. As some of the calculation values will be legitimately zeros and throw a NaN/infinite result. 

This will cause some issues further down the line when trying to fit the ML pipeline so we'll impute the NaNs with a 0 but also create a flag, so we can differentiate between genuine 0s and our replacement values in case it is pertinent later on.

```python
# Add the four 'ratio' values identified above, to assist the analysis to identify possible income or job stagnation.
df['RoleTenureRatio'] = df['YearsInCurrentRole'] / df['YearsAtCompany']
df['ManagerTenureRatio'] = df['YearsWithCurrManager'] / df['YearsAtCompany']
df['IncomeTenureRatio'] = df['MonthlyIncome'] / df['YearsAtCompany']
df['IncomeTotalWorkRatio'] = df['MonthlyIncome'] / df['TotalWorkingYears']  

#amend these ratios to 2 decimal places
df['RoleTenureRatio'] = df['RoleTenureRatio'].round(2)
df['ManagerTenureRatio'] = df['ManagerTenureRatio'].round(2)
df['IncomeTenureRatio'] = df['IncomeTenureRatio'].round(2)
df['IncomeTotalWorkRatio'] = df['IncomeTotalWorkRatio'].round(2)   
df.head(5)
```

```python
# Replace inf and -inf with NaN
df.replace([np.inf, -np.inf], np.nan, inplace=True)
df.head(5)
```

```python
#replace the NaNs with 0s and create flags
ratio_cols = [col for col in df.columns if "ratio" in col.lower()]
df[ratio_cols] = df[ratio_cols].fillna(0)

for col in ratio_cols:
    if not col.endswith("_was_nan"):
        df[f"{col}_was_nan"] = df[col].isna().astype(int)

df.head(10)
```
